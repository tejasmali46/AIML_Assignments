```python
import random


class Environment:

    def __init__(self):
        self.rooms = {
            "Living Room": {
                "dust": random.randint(0, 2),
                "priority": 2,
                "obstacle": random.choice([True, False])
            },
            "Kitchen": {
                "dust": random.randint(0, 2),
                "priority": 3,
                "obstacle": random.choice([True, False])
            },
            "Bedroom": {
                "dust": random.randint(0, 2),
                "priority": 1,
                "obstacle": random.choice([True, False])
            },
            "Balcony": {
                "dust": random.randint(0, 2),
                "priority": 2,
                "obstacle": random.choice([True, False])
            }
        }

    def display(self):
        print("\nROOM CONDITIONS")

        for room, data in self.rooms.items():
            dust = data["dust"]

            #checking room condition
            if dust == 0:
                status = "Clean"
            elif dust == 1:
                status = "Dusty"
            else:
                status = "Very Dirty"

            obstacle = "Yes" if data["obstacle"] else "No"
            print(room, "| Dust:", status, "| Priority:", data["priority"], "| Obstacle:", obstacle)


class SmartVacuum:

    def __init__(self, environment):
        self.env = environment

        #starting from a random room
        self.location = random.choice(list(self.env.rooms.keys()))

        self.battery = 100
        self.dust_bin = 0
        self.bin_capacity = 4
        self.score = 0
        self.rooms_cleaned = 0
        self.movements = 0

    def choose_room(self):
        dirty_rooms = {}

        for room, data in self.env.rooms.items():

            #finding dirty rooms
            if data["dust"] > 0:
                value = data["dust"] * 3 + data["priority"]
                dirty_rooms[room] = value

        #all rooms are clean
        if not dirty_rooms:
            return None

        #choosing the room with more dust
        return max(dirty_rooms, key=dirty_rooms.get)

    def clean(self):
        data = self.env.rooms[self.location]
        dust = data["dust"]

        #room is already clean
        if dust == 0:
            print(self.location, "is already clean.")
            return

        print("\nCleaning", self.location)

        #checking for obstacle
        if data["obstacle"]:
            print("Obstacle found!")
            print("Robot is moving around it...")
            self.battery -= 5

            #obstacle is cleared
            data["obstacle"] = False

        #checking dust bin
        if self.dust_bin + dust > self.bin_capacity:
            print("Dust bin is full.")

            #emptying the dust bin
            self.empty_bin()

        #collecting dust
        self.dust_bin += dust

        #making the room clean
        data["dust"] = 0

        #battery used for cleaning
        self.battery -= dust * 10

        #adding cleaning score
        self.score += dust * 10
        self.rooms_cleaned += 1

        print(" ", self.location, "cleaned successfully.")
        print("Dust collected:", self.dust_bin, "/", self.bin_capacity)

    def move(self, destination):

        #no need to move
        if self.location == destination:
            return

        #moving to next room
        print("\nMoving from", self.location, "to", destination)

        self.location = destination
        self.battery -= 5
        self.movements += 1

        print("Robot reached", destination)

    def recharge(self):

        print("\nBattery is low.")

        #going back for charging
        print("Returning to charging station...")

        self.location = "Living Room"
        self.battery = 100

        print("Battery fully charged.")

    def empty_bin(self):

        print("\nDust bin is full.")

        #going to empty the bin
        print("Returning to Living Room...")

        self.location = "Living Room"
        self.dust_bin = 0

        print("Dust bin emptied.")

    def check_battery(self):

        #battery is getting low
        if self.battery <= 20:
            self.recharge()

    def run(self):

        print("\nSMART PRIORITY VACUUM ACTIVATED")
        print("Starting Location:", self.location)

        self.env.display()

        while True:

            #checking battery
            self.check_battery()

            #finding the next room
            next_room = self.choose_room()

            #all rooms are clean
            if next_room is None:
                print("\n✓ All rooms are clean!")
                break

            #moving to next room
            self.move(next_room)

            #cleaning the room
            self.clean()

            print("\nBattery:", self.battery, "%")
            print("Performance Score:", self.score)

        self.final_report()

    def final_report(self):

        print("\nFINAL REPORT")
        print("Rooms Cleaned:", self.rooms_cleaned)
        print("Movements:", self.movements)
        print("Dust Bin:", self.dust_bin, "/", self.bin_capacity)
        print("Final Battery:", self.battery, "%")
        print("Performance:", self.score)

        self.env.display()


#creating environment
environment = Environment()

#creating vacuum
vacuum = SmartVacuum(environment)

#starting the vacuum
vacuum.run()
```
