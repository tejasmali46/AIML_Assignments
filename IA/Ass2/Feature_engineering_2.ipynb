```python

# SECTION 5: CATEGORICAL ENCODING

import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder

# Create a copy of the dataset
df_encoded = df.copy()

# 1. Label Encoding
label_columns = [
    col for col in ["Transport", "Energy Source"]
    if col in df_encoded.columns
]

for col in label_columns:
    le = LabelEncoder()
    df_encoded[col + "_encoded"] = le.fit_transform(
        df_encoded[col].astype(str)
    )

print("1. Label Encoding Completed")

# 2. Ordinal Encoding for Diet
if "Diet" in df_encoded.columns:
    print("Diet categories:", df_encoded["Diet"].unique())

    # Automatically encode available diet categories
    ordinal_encoder = OrdinalEncoder()
    df_encoded["Diet_ordinal"] = ordinal_encoder.fit_transform(
        df_encoded[["Diet"]].astype(str)
    )

print("2. Ordinal Encoding Completed")

# 3. One-Hot Encoding
one_hot_columns = [
    col for col in ["Vehicle Type", "Social Activity"]
    if col in df_encoded.columns
]

df_encoded = pd.get_dummies(
    df_encoded,
    columns=one_hot_columns,
    dtype=int
)

print("3. One-Hot Encoding Completed")


# 4. Binary Encoding using pandas

if "Recycling" in df_encoded.columns:
    recycling_codes = pd.factorize(
        df_encoded["Recycling"].astype(str)
    )[0]

    binary_width = max(
        1, int(recycling_codes.max()).bit_length()
    )

    for i in range(binary_width):
        df_encoded[f"Recycling_bit_{i}"] = (
            recycling_codes >> i
        ) & 1

print("4. Binary Encoding Completed")

# Display encoded dataset
print("\nEncoded Dataset:")
display(df_encoded.head())

print("\nEncoded Dataset Shape:", df_encoded.shape)
```

```python

# SECTION 6: FEATURE SCALING

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    RobustScaler,
    Normalizer
)

# Select numerical features
numerical_features = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Exclude the target variable from input features
numerical_features = [
    col for col in numerical_features
    if col != "CarbonEmission"
]

print("Numerical Features:", numerical_features)

# Keep only rows with complete numerical values for this demonstration
X = df[numerical_features].copy()
X = X.fillna(X.median())

# 1. Standard Scaling
standard_scaler = StandardScaler()
X_standard = pd.DataFrame(
    standard_scaler.fit_transform(X),
    columns=numerical_features
)

print("\n1. Standard Scaling:")
display(X_standard.head())

# 2. Min-Max Scaling
minmax_scaler = MinMaxScaler()
X_minmax = pd.DataFrame(
    minmax_scaler.fit_transform(X),
    columns=numerical_features
)

print("\n2. Min-Max Scaling:")
display(X_minmax.head())

# 3. Robust Scaling
robust_scaler = RobustScaler()
X_robust = pd.DataFrame(
    robust_scaler.fit_transform(X),
    columns=numerical_features
)

print("\n3. Robust Scaling:")
display(X_robust.head())

# 4. Normalization
normalizer = Normalizer()
X_normalized = pd.DataFrame(
    normalizer.fit_transform(X),
    columns=numerical_features
)

print("\n4. Normalization:")
display(X_normalized.head())

# Compare distributions for one numerical feature
if numerical_features:
    feature = numerical_features[0]

    plt.figure(figsize=(10, 5))
    sns.kdeplot(X_standard[feature], label="StandardScaler")
    sns.kdeplot(X_minmax[feature], label="MinMaxScaler")
    sns.kdeplot(X_robust[feature], label="RobustScaler")
    plt.title(f"Scaling Comparison: {feature}")
    plt.legend()
    plt.show()

print("\nFeature scaling completed successfully.")
```

```python

# SECTION 7: FEATURE ENGINEERING

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import PowerTransformer
from sklearn.decomposition import PCA

# Create a separate copy
df_fe = df.copy()

# 1. Feature Creation
# Create meaningful features from existing lifestyle data

# Total monthly vehicle distance
if "Vehicle Monthly Distance Km" in df_fe.columns:
    df_fe["Vehicle_Distance_Per_Day"] = (
        df_fe["Vehicle Monthly Distance Km"] / 30
    )

# Combined waste generated and recycled
if "Waste Generated Kg" in df_fe.columns:
    df_fe["Waste_Generated_Per_Day"] = (
        df_fe["Waste Generated Kg"] / 30
    )

print("1. Feature Creation Completed")

# 2. Feature Transformation using Yeo-Johnson
if "CarbonEmission" in df_fe.columns:
    transformer = PowerTransformer(method="yeo-johnson")

    df_fe["CarbonEmission_Transformed"] = transformer.fit_transform(
        df_fe[["CarbonEmission"]]
    )

print("2. Feature Transformation Completed")

# 3. Feature Extraction using PCA
pca_features = df_fe.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Do not include the target or transformed target in PCA
pca_features = [
    col for col in pca_features
    if col not in [
        "CarbonEmission",
        "CarbonEmission_Transformed"
    ]
]

if len(pca_features) >= 2:
    X_pca = df_fe[pca_features].copy()
    X_pca = X_pca.fillna(X_pca.median())

    from sklearn.preprocessing import StandardScaler

    scaler_pca = StandardScaler()
    X_scaled_pca = scaler_pca.fit_transform(X_pca)

    pca = PCA(n_components=1)
    df_fe["Lifestyle_PCA"] = pca.fit_transform(X_scaled_pca)

    print(
        "3. PCA Completed. Explained variance:",
        round(pca.explained_variance_ratio_[0], 4)
    )

# 4. Outlier Detection and Handling using IQR
if "CarbonEmission" in df_fe.columns:
    Q1 = df_fe["CarbonEmission"].quantile(0.25)
    Q3 = df_fe["CarbonEmission"].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_fe[
        (df_fe["CarbonEmission"] < lower_bound) |
        (df_fe["CarbonEmission"] > upper_bound)
    ]

    print("\nNumber of potential target outliers:", len(outliers))

    plt.figure(figsize=(8, 5))
    sns.boxplot(y=df_fe["CarbonEmission"])
    plt.title("Carbon Emission Before Outlier Handling")
    plt.show()

    # Create a capped target column for demonstration
    df_fe["CarbonEmission_Capped"] = (
        df_fe["CarbonEmission"].clip(
            lower=lower_bound,
            upper=upper_bound
        )
    )

    plt.figure(figsize=(8, 5))
    sns.boxplot(y=df_fe["CarbonEmission_Capped"])
    plt.title("Carbon Emission After IQR Capping")
    plt.show()

print("\nFeature Engineering Completed")

print("\nUpdated Dataset:")
display(df_fe.head())

print("\nUpdated Dataset Shape:", df_fe.shape)
```

```python

# SECTION 8: FINAL PREPROCESSED DATASET

import pandas as pd
from sklearn.preprocessing import StandardScaler

# Start from the original dataset
final_df = df.copy()

# Separate target variable
target = "CarbonEmission"

# Handle missing categorical values
categorical_cols = final_df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

for col in categorical_cols:
    final_df[col] = final_df[col].fillna("Unknown")

# Handle missing numerical values
numerical_cols = final_df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

for col in numerical_cols:
    if col != target:
        final_df[col] = final_df[col].fillna(
            final_df[col].median()
        )

# Encode categorical features using one-hot encoding
final_df = pd.get_dummies(
    final_df,
    columns=categorical_cols,
    dtype=int
)

# Separate features and target
X = final_df.drop(columns=[target])
y = final_df[target]

# Scale numerical input features only
numeric_input_cols = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

scaler = StandardScaler()

if numeric_input_cols:
    X[numeric_input_cols] = scaler.fit_transform(
        X[numeric_input_cols]
    )

# Combine features and target
final_preprocessed_df = X.copy()
final_preprocessed_df[target] = y

# Check final dataset
print("Final Dataset Shape:", final_preprocessed_df.shape)

print("\nMissing Values:")
print(final_preprocessed_df.isnull().sum().sum())

print("\nFinal Dataset Preview:")
display(final_preprocessed_df.head())

# Export final dataset
output_file = "final_carbon_footprint_dataset.csv"

final_preprocessed_df.to_csv(
    output_file,
    index=False
)

print("\nDataset saved successfully:", output_file)

# Download CSV in Google Colab
from google.colab import files
files.download(output_file)
```
